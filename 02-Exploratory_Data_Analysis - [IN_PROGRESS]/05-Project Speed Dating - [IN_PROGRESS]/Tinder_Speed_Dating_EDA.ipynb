3. Nettoyage du dataset

```python
# Vérifier les valeurs manquantes
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values_percentage = (missing_values / len(df)) * 100
missing_values_percentage

# Garder uniquement les colonnes avec moins de 40% de valeurs manquantes
cols_to_keep = missing_values_percentage[missing_values_percentage < 40].index
df_clean = df[cols_to_keep]
df_clean

# Vérifier la taille après nettoyage
df_clean.shape
```

```python
# Préférences initiales des participants par genre
prefs_cols = ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1', 'shar1_1']
prefs_by_gender = df.groupby('gender')[prefs_cols].mean().T

# Préparation des données pour le graphique interactif
prefs_by_gender_plot = df.groupby('gender')[prefs_cols].mean().T.reset_index()
prefs_by_gender_plot.columns = ['Critère', 'Femmes', 'Hommes']


import plotly.express as px

# Graphique interactif des préférences selon le genre avec couleurs personnalisées
fig_prefs = px.bar(
    prefs_by_gender_plot, 
    x='Critère', 
    y=['Femmes', 'Hommes'],
    title="Importance moyenne des critères initiaux selon le genre",
    labels={'value': 'Points moyens attribués', 'Critère': 'Critères'},
    barmode='group',
    color_discrete_sequence=['#E75480', '#4169E1']  # Couleurs personnalisées : rose pâle et bleu clair
)

fig_prefs.update_layout(
    xaxis_tickangle=-45,
    yaxis=dict(
        gridcolor='grey',    # Grille plus discrète sur fond sombre
        color='orange'       # Couleur des chiffres sur l'axe Y en orange
    ),
    xaxis=dict(
        color='orange'       # Couleur des critères sur l'axe X en orange
    ),
    legend_title_text='Genre',  # Titre de légende
    legend=dict(
        font=dict(color='gold'),  # Couleur du texte de la légende
        title_font=dict(color='gold')
    ),
    title_font=dict(size=24, family='Arial', color='gold'),  # Titre principal en doré
    plot_bgcolor='rgba(0,0,0,0)',  # Fond transparent
    paper_bgcolor='rgba(0,0,0,0)'  # Fond transparent
)

fig_prefs.show()
```

### Interprétation des résultats

#### Question de recherche
**Quels sont les attributs les moins désirables chez un partenaire masculin ? Est-ce différent chez une partenaire féminine ?**

---

#### 1. Résultats par genre

##### Femmes
- Accordent relativement **moins d'importance** à :
  - **Ambition** (amb)
  - **Partage d'intérêts / hobbies** (shar)
- L'**attractivité physique**, la **sincérité**, l'**intelligence** et le **côté amusant (fun)** restent les critères les plus valorisés.

##### Hommes
- Accordent significativement **moins d'importance** à :
  - **Ambition** (amb)
  - **Partage d'intérêts / hobbies** (shar)
- Leur préférence est **très concentrée** sur l'**attractivité physique**, suivie par le **fun**.

---

#### 2. Analyse comparative
- **Chez les femmes** : Les critères moins importants sont **ambition** et **partage d'intérêts**, mais les écarts entre les critères sont **moins marqués**.  
- **Chez les hommes** : Les critères **ambition** et **partage d'intérêts** sont aussi **moins valorisés**, mais la **priorité donnée à l'apparence physique est beaucoup plus marquée** que chez les femmes.

---

#### Conclusion
 Les **attributs les moins désirables** sont **l'ambition** et **le partage d'intérêts/hobbies**, aussi bien pour les partenaires masculins que féminins.  
 **Cependant**, **les hommes** affichent une **préférence beaucoup plus marquée pour l'apparence physique**, tandis que **les femmes** sont **plus équilibrées** dans l'importance accordée aux divers critères.



```python
# Corrélation avec 'match'
attributes_cols = ['attr', 'sinc', 'intel', 'fun', 'amb', 'shar', 'like', 'match']
corr_matrix = df[attributes_cols].corr()

fig_corr = px.imshow(corr_matrix,
                     text_auto='.2f',
                     color_continuous_scale='YlOrRd', 
                     title='Corrélation entre les attributs évalués et le succès des rencontres')

fig_corr.update_layout(
    margin=dict(l=20, r=20, t=60, b=20),
    title_font=dict(size=24, family='Arial', color='gold'),
    xaxis=dict(showgrid=False, side='bottom', ticks='', color='orange'),
    yaxis=dict(showgrid=False, ticks='',color='orange'),
    plot_bgcolor='rgba(0,0,0,0)',
    paper_bgcolor='rgba(0,0,0,0)',
    coloraxis_colorbar=dict(
        title='Corrélation',
        thickness=20,
        tickformat='.2f',
        tickfont=dict(color='gold'), # Ticks de la barre en or
        title_font=dict(color='gold'),  # Titre de la barre en or
    )
)

fig_corr.update_xaxes(tickangle=-45)
fig_corr.update_yaxes(autorange='reversed')

fig_corr.show()
```
