2. Read the files and create the 4 corresponding Dataframes


```python
import pandas as pd
```

```python

df_Surface = pd.read_csv(r"house_market\superficie.csv",header=None)        # header = None -> ne prends pas la première ligne comme ligne d'en-tête
df_Bathrooms = pd.read_csv(r"house_market\nombre_de_sdb.csv",header=None)   # r"xxx" -> texte brut. Cela évite les erreurs comme \n (retour à la ligne) et autres.
df_Rooms = pd.read_csv(r"house_market\nombre_de_chambres.csv",header=None)
df_House_prices = pd.read_csv(r"house_market\house_price.csv",index_col=0)  # index_col = 0 -> la colonne 0 sert d'index.
```

```python
print("Surface:")
display(df_Surface.head())
print()
print("Bathrooms:")
display(df_Bathrooms.head())
print()
print("Rooms:")
display(df_Rooms.head())
print()
print("House prices:")
display(df_House_prices.head())
```

```python
df_Surface.columns.to_list()
```

3. Give a column name to the three datasets that don't have a header in the source file

```python
df_Surface.columns = ["id","surface"]
df_Bathrooms.columns = ["id","bathrooms"]
df_Rooms.columns = ["id","rooms"]
```

```python
print("Surface:")
display(df_Surface.head())
print()
print("Bathrooms:")
display(df_Bathrooms.head())
print()
print("Rooms:")
display(df_Rooms.head())
print()
print("House prices:")
display(df_House_prices.head())
```

4. Make an inner join of these four datasets into a single dataframe.

```python
df_total = pd.merge(df_Surface,df_Bathrooms, on='id')
df_total = pd.merge(df_total,df_Rooms, on='id')
df_total = pd.merge(df_total,df_House_prices, on='id')
df_total.head()
```

5. What is the average surface area of the houses in our dataset?

```python
print(f"The average surface area is : {df_total['surface'].mean()}")
```

6. What is the median and average number of rooms?

```python
print(f"The average number of room is : {df_total['rooms'].mean()}, and the median number is : {df_total['rooms'].median()}")
```

7. What is the average cost of a house?

```python
print(f"The average cost of a house is {df_total["house_price"].mean():,.2f}".replace(","," ") +" $")
```

8. What is the average cost of a house, depending on the number of rooms it has?

```python
print("Rooms")
df_total.groupby("rooms")["house_price"].mean().round(2)
```

10. The average cost per room doesn't tell us much. Let's try to categorize by size.

  - Create a new column in your dataset that we'll call home_size
Create five categories that respectively correspond to:
      - "very large" $=$ "a house larger than 25,000 sqrt_feet"
      - "large" $=$ "a house between 20,000 and 25,000 sqrt_feet"
      - "medium" $=$ "a house between 15,000 and 20,000 sqrt_feet"
      - "small" $=$ "a house between 10,000 and 15,000 sqrt_feet"
      - "very small" $=$ a house less than 10,000 sqrt_feet"

Fill in the corresponding rows in the house_size column.

```python
df_total["house_size"] = df_total["surface"].apply(
    lambda x: "very small" if x < 10000 else  
              "small" if x < 15000 else
              "medium" if x < 20000 else
              "large" if x < 25000 else
              "very large"
)
df_total.head()
```

11. What is the average cost of a house depending on its size category?


```python
df_total.groupby("house_size")["house_price"].mean().round(2).sort_values(ascending=False)
```
