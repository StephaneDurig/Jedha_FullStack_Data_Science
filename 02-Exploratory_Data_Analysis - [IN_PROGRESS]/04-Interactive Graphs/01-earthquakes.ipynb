## Visualizing earthquakes

Let's use Plotly to make an exploratory analysis of a dataset that gathers data from various incidents around the world, and in particular lots of earthquakes that have occured since 1965 !

The dataset is available on Kaggle, you can download it and you'll find a description here :  https://www.kaggle.com/usgs/earthquake-database

## Beginning with the dataset

1. Import pandas and the different packages of Plotly

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
# pio.renderers.default = "svg" # this line must be commented if working on colab

```

```python
pd.options.display.max_columns = None
```

2. Read the file and show the first rows as well as basic statistics about the dataset

```python
df = pd.read_csv("earthquakes.csv")
df.head()
```

```python
df.describe(include='all')
```

3. In the following, we will use a lot the `Date` column. To avoid problems with your visualizations, use pandas to convert this column into `Datetime` type :

```python
df['Date'] =pd.to_datetime(df['Date'])
df.head()
```

## Exploring the number of observations over time

4. Make a simple plot with plotly express to display the number of observed events as a function of time

```python
fig = px.histogram(df,x='Date')

fig.update_layout(width=700, height=500)

fig.show()
```

If the `Date` column was converted into `Datetime`, plotly handles it and has automatically set the number of bins such that the counts are made by year. This makes the histogram easily readable.

In the following, you will create customized plots with plotly.graph_objects to display more accurate information about the number of events.

5. With plotly.graph_objects, create a histogram with a range slider such that you can visualize the daily number of events

```python
go.Figure(
    data = go.Histogram(
        x = df['Date'],
        nbinsx = df['Date'].nunique()   # nunique() -> le nombre de barre = le nombre de jour unique.
    ),
    layout = go.Layout(
        width = 700,
        height = 500,
        title = go.layout.Title(text = "Number of observations per day", x = 0.5),
        xaxis = go.layout.XAxis(title = 'Date', rangeslider = go.layout.xaxis.Rangeslider(visible = True))
    )
)
```

6. We would like to check if there's some seasonality in the number of events. Create new columns in the dataset with the id of the month and year. Then use plotly's `make_subplots` function to display the monthly number of observations over a given year, independently for each of the five last years (so, there will be 5 independent histograms)

```python
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df.head()
```

```python
print(f"La dernière année du dataset est: {df['year'].max()}")
```

```python
from plotly.subplots import make_subplots

# Liste des années
annees = [2012,2013,2014,2015,2016]
n = len(annees)

# Créer les subplots : 1 colonne, n lignes (sans titres)
fig = make_subplots(rows=n, cols=1, shared_xaxes = True, vertical_spacing=0.02)

# Ajouter les histogrammes par année, tous avec légende visible
for i, annee in enumerate(annees):
    data_annee = df[df['year'] == annee]
    fig.add_trace(
        go.Histogram(
            x=data_annee['month'],
            name=str(annee),       # Le nom apparaîtra dans la légende
        ),
        row=i+1, col=1
    )

    # Appliquer la même échelle pour chaque axe Y
    fig.update_yaxes(range=[0,85], showticklabels=True, row=i+1, col=1)
    
# Mise en page
fig.update_layout(
    height = 180 * n,
    width = 800,
    bargap=0.05, # Gère l'écartement entre les barres de l'histogramme
    title_text="Histogrammes mensuels des observations par année",
    title_x=0.5
)

fig.show()
```

7. Now, we would like to allow the user to choose a specific year among the five last years, and display the daily number of observations fot this year only. Use the `Updatemenu` and `Button` classes together with `add_trace` to create an interactive visualization.

```python
import plotly.graph_objects as go

# Liste des années disponibles
annees = [2012, 2013, 2014, 2015, 2016]

# Créer la figure
fig = go.Figure()

# Ajouter un histogramme par année (seul le premier est visible au départ)
for i, annee in enumerate(annees):
    data_annee = df[df['year'] == annee]
    fig.add_trace(
        go.Histogram(
            x=data_annee['Date'],
            nbinsx = data_annee['Date'].nunique(),
            name=str(annee),
            visible=(i == 0)
        )
    )

# Créer les boutons pour chaque année
buttons = []
for i, annee in enumerate(annees):
    visibility = [j == i for j in range(len(annees))]
    buttons.append(
        dict(
            label=str(annee),
            method="update",
            args=[{"visible": visibility},
                  {"title": f"Histogramme pour l'année {annee}"}]
        )
    )

# Ajouter le menu dans le layout
fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=buttons,
            x=1.1,
            y=1.15
        )
    ],
    width=700,
    height=500,
    title="Histogramme pour l'année 2012",
    xaxis_title="Mois",
    yaxis_title="Nombre d'observations",
    )

fig.show()
```

10. Now, we'd like to explore the evolution of the magnitudes as a function of time :


- Create a dataframe containing the mean magnitudes per day
- Then use this dataset to display the mean magnitudes per day, with a range slider allowing the user to navigate among the dates
- Add a reference line showing the value of the mean magnitude computed over the whole dataset


```python
df_magnitude_by_date = df_earthquakes.groupby('Date')['Magnitude'].mean().reset_index(name='Magnitude')
df_magnitude_by_date.head()
```

```python
fig = go.Figure(
        data = go.Scatter(
            x = df_magnitude_by_date['Date'],
            y = df_magnitude_by_date['Magnitude'],
        ),
        layout = go.Layout(
            width = 700,
            height = 500,
            title = go.layout.Title(text = "Mean magnitude per day", x = 0.5),
            xaxis = go.layout.XAxis(title = 'Date', rangeslider = go.layout.xaxis.Rangeslider(visible = True))
        )
)

fig.show()
```

**Méthode avec .add_trace:**  
  
Crée une donnée de série :  
* Visible dans le graph  
* Visible dans le range slider  
* Inclus dans la légende (si showlegend = True)  
* ...

```python
# Moyenne globale de la magnitude
moyenne_magnitude = df_magnitude_by_date['Magnitude'].mean()

fig.add_trace(
    go.Scatter(
        x = df_magnitude_by_date['Date'],
        y = [moyenne_magnitude] * len(df_magnitude_by_date),
        mode = 'lines',
        line = dict(color = "red")
    )
)
fig.update_layout(showlegend = False)

fig.show()
```

**Méthode avec shapes:**  
  
Éléments décoratifs, non liés aux données :  
* Dessiné par dessus le graph  
* Non Visible dans le range slider  
* Non Visible dans la légende  
* ...

```python
# Moyenne globale de la magnitude
moyenne_magnitude = df_magnitude_by_date['Magnitude'].mean()

# Création du graphique
fig = go.Figure(
    data = go.Scatter(
        x = df_magnitude_by_date['Date'],
        y = df_magnitude_by_date['Magnitude'],
    ),
    layout = go.Layout(
        width = 700,
        height = 500,
        title = go.layout.Title(text = "Mean magnitude per day", x = 0.5),
        xaxis = go.layout.XAxis(title = 'Date', rangeslider = go.layout.xaxis.Rangeslider(visible = True)),
        shapes = [
            # Ligne horizontale pour la moyenne
            dict(
                type="line",
                xref="paper",  # toute la largeur du graphique
                x0=0,
                x1=1,
                yref="y",
                y0 = moyenne_magnitude,
                y1 = moyenne_magnitude,
                line = dict(color = "red")
            )   
        ]
    )
)

fig.show()

```

11. Use plotly express' `scatter_mapbox` to display the earthquakes on a map. Change the color of the markers depending on the value of the magnitude. Use the documentation and [this page](https://plotly.com/python/builtin-colorscales/) to find a suitable colorscale.

```python
#  px.scatter_map?
```

```python
fig = px.scatter_map(
    df_earthquakes,
    lat = "Latitude",
    lon = "Longitude",
    color = "Magnitude",
    map_style = "open-street-map",
    zoom = 0.5,
    color_continuous_scale = 'Reds'
    )

fig.update_layout(width=700, height=500)

fig.show()
```

12. Let's make the map animated ! Add some arguments in `scatter_mapbox` to create an animation displaying the earthquakes year by year

```python
fig = px.scatter_map(
    df_earthquakes,
    lat = "Latitude",
    lon = "Longitude",
    color = "Magnitude",
    map_style = "open-street-map",
    zoom = 0.5,
    color_continuous_scale = 'Reds',
    range_color = [5.0,10.0],
    animation_frame = 'year'
    )

fig.update_layout(width=700, height=500)

fig.show()
```
